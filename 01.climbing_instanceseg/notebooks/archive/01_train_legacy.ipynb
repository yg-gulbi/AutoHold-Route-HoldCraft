```python
!pip install ultralytics
from ultralytics import YOLO

# 1. Base 모델 초기화
model_path = 'yolo11m-seg.pt'
yaml_data = '/workspace/hold_seg_data-1/data.yaml'
model = YOLO(model_path)

# 2. 신규 학습 실행
results = model.train(
    data= yaml_data,
    epochs=100,                        # 총 학습 에폭 수
    imgsz=640,                         # 이미지 크기
    batch=64,                          # 배치 사이즈 (VRAM 부족 시 36으로 변경)
    name='yolo11_climbing_seg',  # 결과 저장 폴더명
    project='/workspace/runs/train',
    device=0,                          # GPU 번호
    patience=20,                       # 조기 종료 설정
    save=True,
    resume=False                       # 처음부터 학습 진행
)
print("end")
```
